```python
import pandas as pd
```

```python
IMAGES_DIR = '../images'
IMAGES_DATA_DIR = '../images/data'
BASE_LOSS_NAME = 'Loss'
BASE_ACCURACY_NAME = 'Accuracy'
METRICS = [BASE_LOSS_NAME, BASE_ACCURACY_NAME]
MODES = ['train', 'val']
```

```python
POSTFIX = 'vanilla'
```

```python
dfs = {}
for metric in METRICS:
    for mode in MODES:
        key = f"{metric}_{mode}"
        dfs[key] = pd.read_csv(f"{IMAGES_DATA_DIR}/{metric}_{mode}_{POSTFIX}.csv")
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# --- Left subplot: Loss ---
ax = axes[0]
for mode in MODES:
    key = f"{BASE_LOSS_NAME}_{mode}"
    ax.plot(dfs[key]["Step"], dfs[key]["Value"], label=mode.capitalize())
ax.set_title("Loss")
ax.set_xlabel("Step")
ax.set_ylabel("Value")
ax.grid(True)
ax.legend()

# --- Right subplot: Accuracy ---
ax = axes[1]
for mode in MODES:
    key = f"{BASE_ACCURACY_NAME}_{mode}"
    ax.plot(dfs[key]["Step"], dfs[key]["Value"], label=mode.capitalize())
ax.set_title("Accuracy")
ax.set_xlabel("Step")
ax.set_ylabel("Value")
ax.grid(True)
ax.legend()

# Overall title
plt.suptitle(f"{POSTFIX} Results", fontsize=18, fontweight='bold')

plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```
